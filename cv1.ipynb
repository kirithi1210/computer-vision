    import cv2
    import matplotlib.pyplot as plt

    # Load the image in RGB (OpenCV loads images in BGR by default, so we need to convert it to RGB)
    image_bgr = cv2.imread('lenna.jpg')
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    # Separate the image into its Red, Green, and Blue channels
    r, g, b = cv2.split(image_rgb)

    # Apply Histogram Equalization to each channel
    r_eq = cv2.equalizeHist(r)
    g_eq = cv2.equalizeHist(g)
    b_eq = cv2.equalizeHist(b)

    # Merge the equalized channels back into a single image
    equalized_image_rgb = cv2.merge([r_eq, g_eq, b_eq])

    # Create a figure to display images and histograms
    plt.figure(figsize=(12, 8))

    # Plot the original image and its histogram
    plt.subplot(2, 2, 1)
    plt.imshow(image_rgb)
    plt.title('Original Image (RGB)')
    plt.axis('off')

    # Plot the original image histogram for each channel
    plt.subplot(2, 2, 2)
    plt.hist(r.ravel(), bins=256, range=(0, 256), color='red', alpha=0.6, label='Red')
    plt.hist(g.ravel(), bins=256, range=(0, 256), color='green', alpha=0.6, label='Green')
    plt.hist(b.ravel(), bins=256, range=(0, 256), color='blue', alpha=0.6, label='Blue')
    plt.title('Original Image Histogram')
    plt.legend()

    # Plot the equalized image and its histogram
    plt.subplot(2, 2, 3)
    plt.imshow(equalized_image_rgb)
    plt.title('Equalized Image (RGB)')
    plt.axis('off')

    # Plot the equalized image histogram for each channel
    plt.subplot(2, 2, 4)
    plt.hist(r_eq.ravel(), bins=256, range=(0, 256), color='red', alpha=0.6, label='Red')
    plt.hist(g_eq.ravel(), bins=256, range=(0, 256), color='green', alpha=0.6, label='Green')
    plt.hist(b_eq.ravel(), bins=256, range=(0, 256), color='blue', alpha=0.6, label='Blue')
    plt.title('Equalized Image Histogram')
    plt.legend()

    # Show the images and histograms
    plt.tight_layout()
    plt.show()

    import cv2
    import numpy as np
    import matplotlib.pyplot as plt

    # Load the image
    image = cv2.imread('lenna.jpg')  # Replace with your image path

    # Get the original image dimensions
    original_height, original_width, channels = image.shape

    # Define new dimensions for resizing (for example, resize to 800x600)
    new_width = 800
    new_height = 600

    # Create a new empty image with the resized dimensions
    resized_image = np.zeros((new_height, new_width, channels), dtype=np.uint8)

    # Calculate the scaling factors
    x_scale = original_width / new_width
    y_scale = original_height / new_height

    # Iterate through each pixel of the resized image
    for i in range(new_height):
        for j in range(new_width):
            # Find the corresponding pixel in the original image
            orig_x = int(j * x_scale)
            orig_y = int(i * y_scale)
            
            # Copy the color from the original image to the resized image
            resized_image[i, j] = image[orig_y, orig_x]

    # Convert the images to RGB (for displaying in Matplotlib)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    resized_image_rgb = cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)

    # Display the original and resized images
    plt.figure(figsize=(12, 6))

    # Original Image
    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title(f'Original Image\nDimensions: {original_width}x{original_height}')
    plt.axis('off')

    # Resized Image
    plt.subplot(1, 2, 2)
    plt.imshow(resized_image_rgb)
    plt.title(f'Resized Image\nDimensions: {new_width}x{new_height}')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
